# 1. Diagnóstico do colapso

Esta seção produz evidências para investigar por que a quantização integral do MobileViT-XXS perdeu acurácia. A análise utiliza as estatísticas coletadas durante a calibração estratificada, o checkpoint FP32 e os resultados reais do LiteRT.

Os gráficos são exportados em PNG, com 300 dpi, e em PDF vetorial.

````python
import json
from pathlib import Path

import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import torch

sns.set_theme(style="whitegrid", context="paper")
mpl.rcParams.update({
    "figure.dpi": 120,
    "savefig.dpi": 300,
    "font.size": 10,
    "axes.titlesize": 11,
    "axes.labelsize": 10,
    "legend.fontsize": 9,
})
````

````python
CURRENT_DIR = Path.cwd()

if (CURRENT_DIR / "gerar_plots_para_o_artigo.ipynb").is_file():
    ARTICLE_DIR = CURRENT_DIR
elif (CURRENT_DIR / "artigo_experimentos" / "gerar_plots_para_o_artigo.ipynb").is_file():
    ARTICLE_DIR = CURRENT_DIR / "artigo_experimentos"
else:
    raise FileNotFoundError("Execute o notebook a partir de butterflies_austria_edge_vision ou artigo_experimentos")

PROJECT_ROOT = ARTICLE_DIR.parent
EXPERIMENT_DIR = (
    PROJECT_ROOT
    / "pytorch_mobilevit_xxs"
    / "MobileVIT"
    / "MobileVIT-XXS"
    / "butterflies_austria_MobileVIT_apple_baseline_v2_esp_verify"
)
RESULTS_DIR = EXPERIMENT_DIR / "test1_results"
CHECKPOINT_PATH = (
    ARTICLE_DIR
    / "artigo_experimentos_MobileVIT_v2_esp_verify"
    / "best_mobilevit_xx_small_v0.pt"
)
FIGURES_DIR = ARTICLE_DIR / "figures_diagnostico_colapso"
TABLES_DIR = ARTICLE_DIR / "tables_diagnostico_colapso"

FIGURES_DIR.mkdir(exist_ok=True)
TABLES_DIR.mkdir(exist_ok=True)

required_paths = [
    RESULTS_DIR / "channel_statistics.csv",
    RESULTS_DIR / "group_statistics.csv",
    RESULTS_DIR / "group_configuration.json",
    RESULTS_DIR / "results.csv",
    CHECKPOINT_PATH,
]

missing_paths = [path for path in required_paths if not path.is_file()]
if missing_paths:
    raise FileNotFoundError("\n".join(str(path) for path in missing_paths))

print(f"Results: {RESULTS_DIR}")
print(f"Checkpoint: {CHECKPOINT_PATH}")
print(f"Figures: {FIGURES_DIR}")
````

## 1.1 Evidência de inadequação da escala por tensor

Para cada canal `c`, a escala ideal aproximada é calculada por:

```text
s_c = (max(x_c) - min(x_c)) / 255
```

Na quantização por tensor, todos os canais compartilham

```text
s_T = (max_c(max(x_c)) - min_c(min(x_c))) / 255
```

O número aproximado de códigos INT8 disponíveis para um canal é

```text
N_c = min(256, floor((max(x_c) - min(x_c)) / s_T) + 1)
```

Valores baixos de `N_c` indicam que um canal utiliza apenas uma pequena parte dos 256 níveis disponíveis.

````python
channel_statistics = pd.read_csv(RESULTS_DIR / "channel_statistics.csv")
group_statistics = pd.read_csv(RESULTS_DIR / "group_statistics.csv")
results = pd.read_csv(RESULTS_DIR / "results.csv")

with (RESULTS_DIR / "group_configuration.json").open() as file:
    group_configuration = {
        int(block): groups
        for block, groups in json.load(file).items()
    }

expected_columns = {
    "block",
    "block_name",
    "channel",
    "minimum",
    "maximum",
    "channel_scale",
    "tensor_scale",
    "approximate_int8_codes",
}
missing_columns = expected_columns.difference(channel_statistics.columns)
if missing_columns:
    raise ValueError(f"Missing columns: {sorted(missing_columns)}")

selected_blocks = sorted(group_configuration)
channel_statistics["selected_for_grouping"] = channel_statistics["block"].isin(selected_blocks)

block_summary = (
    channel_statistics.groupby(["block", "block_name"], as_index=False)
    .agg(
        channel_count=("channel", "count"),
        minimum_channel_scale=("channel_scale", "min"),
        maximum_channel_scale=("channel_scale", "max"),
        median_channel_scale=("channel_scale", "median"),
        minimum_approximate_codes=("approximate_int8_codes", "min"),
        median_approximate_codes=("approximate_int8_codes", "median"),
        channels_with_at_most_16_codes=("approximate_int8_codes", lambda values: int((values <= 16).sum())),
        channels_with_at_most_32_codes=("approximate_int8_codes", lambda values: int((values <= 32).sum())),
    )
)
block_summary["scale_ratio"] = (
    block_summary["maximum_channel_scale"]
    / block_summary["minimum_channel_scale"]
)
block_summary["selected_for_grouping"] = block_summary["block"].isin(selected_blocks)
block_summary.to_csv(TABLES_DIR / "block_activation_summary.csv", index=False)

display(
    block_summary[
        [
            "block",
            "channel_count",
            "scale_ratio",
            "minimum_approximate_codes",
            "median_approximate_codes",
            "channels_with_at_most_16_codes",
            "selected_for_grouping",
        ]
    ].round(2)
)
````

````python
def save_figure(figure, name):
    png_path = FIGURES_DIR / f"{name}.png"
    pdf_path = FIGURES_DIR / f"{name}.pdf"
    figure.savefig(png_path, bbox_inches="tight", dpi=300)
    figure.savefig(pdf_path, bbox_inches="tight")
    plt.show()
    print(f"Saved: {png_path}")
    print(f"Saved: {pdf_path}")
````

````python
figure, axes = plt.subplots(1, 2, figsize=(11, 4.2))

colors = [
    "#C44E52" if selected else "#4C72B0"
    for selected in block_summary["selected_for_grouping"]
]

bars = axes[0].bar(
    block_summary["block"].astype(str),
    block_summary["scale_ratio"],
    color=colors,
)

axes[0].set_yscale("log")
axes[0].set_xlabel("Inverted residual block")
axes[0].set_ylabel("Max/min channel-scale ratio")
axes[0].set_title("(a) Activation-scale imbalance")
axes[0].grid(axis="x", visible=False)

for bar, value in zip(bars, block_summary["scale_ratio"]):
    axes[0].text(
        bar.get_x() + bar.get_width() / 2,
        value * 1.08,
        f"{value:.0f}×",
        ha="center",
        va="bottom",
        fontsize=8,
    )

blocks = sorted(channel_statistics["block"].unique())
positions = np.arange(len(blocks))
code_values = [
    channel_statistics.loc[
        channel_statistics["block"] == block,
        "approximate_int8_codes",
    ].to_numpy()
    for block in blocks
]

axes[1].boxplot(
    code_values,
    positions=positions,
    widths=0.55,
    patch_artist=True,
    showfliers=False,
    boxprops={
        "facecolor": "#9EC1E6",
        "edgecolor": "#2F5597",
    },
    medianprops={
        "color": "#C44E52",
        "linewidth": 1.5,
    },
    whiskerprops={
        "color": "#2F5597",
    },
    capprops={
        "color": "#2F5597",
    },
)

random_generator = np.random.default_rng(42)

for position, values in zip(positions, code_values):
    jitter = random_generator.uniform(-0.18, 0.18, size=len(values))
    axes[1].scatter(
        position + jitter,
        values,
        color="#2F5597",
        alpha=0.45,
        s=10,
    )

axes[1].set_xticks(positions)
axes[1].set_xticklabels(blocks)
axes[1].axhline(
    256,
    color="#555555",
    linestyle="--",
    linewidth=1,
    label="Full INT8 range",
)
axes[1].set_yscale("log", base=2)
axes[1].set_ylim(1, 320)
axes[1].set_xlabel("Inverted residual block")
axes[1].set_ylabel("Approximate codes used per channel")
axes[1].set_title("(b) Effective INT8-code utilization")
axes[1].legend(frameon=False, loc="lower right")

figure.tight_layout()
save_figure(
    figure,
    "figure_1_activation_scale_diagnosis",
)
````

## 1.2 Efeito do agrupamento estrutural

A configuração g8_4_4 divide os canais dos blocos 1, 2 e 5 em grupos de magnitudes semelhantes. O gráfico seguinte ordena os canais por escala e identifica os grupos usados para criar as ramificações independentes.

````python
group_lookup = {
    (block, int(channel)): group_index + 1
    for block, groups in group_configuration.items()
    for group_index, channels in enumerate(groups)
    for channel in channels
}

selected_channel_statistics = channel_statistics[
    channel_statistics["block"].isin(selected_blocks)
].copy()
selected_channel_statistics["group"] = [
    group_lookup[(int(block), int(channel))]
    for block, channel in zip(
        selected_channel_statistics["block"],
        selected_channel_statistics["channel"],
    )
]

worst_group_ratio = (
    group_statistics.groupby("block", as_index=False)["scale_ratio"]
    .max()
    .rename(columns={"scale_ratio": "maximum_within_group_ratio"})
)
grouping_summary = (
    block_summary[block_summary["block"].isin(selected_blocks)][
        ["block", "scale_ratio"]
    ]
    .rename(columns={"scale_ratio": "original_tensor_ratio"})
    .merge(worst_group_ratio, on="block", how="left")
)
grouping_summary["ratio_reduction"] = (
    grouping_summary["original_tensor_ratio"]
    / grouping_summary["maximum_within_group_ratio"]
)
grouping_summary.to_csv(TABLES_DIR / "grouping_scale_summary.csv", index=False)

display(grouping_summary.round(2))
````

````python
figure, axes = plt.subplots(
    len(selected_blocks),
    1,
    figsize=(10, 7.5),
    sharex=False,
)

palette = sns.color_palette("tab10", 10)

for axis, block in zip(axes, selected_blocks):
    block_data = (
        selected_channel_statistics[
            selected_channel_statistics["block"] == block
        ]
        .sort_values("channel_scale")
        .reset_index(drop=True)
    )
    block_data["rank"] = np.arange(1, len(block_data) + 1)

    for group, group_data in block_data.groupby("group", sort=True):
        axis.scatter(
            group_data["rank"],
            group_data["channel_scale"],
            s=32,
            color=palette[(int(group) - 1) % len(palette)],
            label=f"Group {int(group)}",
        )

    boundaries = np.flatnonzero(
        np.diff(block_data["group"].to_numpy()) != 0
    ) + 1.5
    for boundary in boundaries:
        axis.axvline(boundary, color="#777777", linestyle=":", linewidth=0.8)

    group_count = len(group_configuration[block])
    axis.set_yscale("log")
    axis.set_ylabel("Channel scale")
    axis.set_title(f"Block {block}: {group_count} groups")
    axis.legend(
        frameon=False,
        ncol=min(group_count, 8),
        loc="upper left",
        fontsize=7,
    )

axes[-1].set_xlabel("Channel rank after sorting by activation scale")
figure.suptitle("Channel partition used by g8_4_4", y=1.01)
figure.tight_layout()
save_figure(figure, "figure_2_g8_4_4_channel_partition")
````

## 1.3 Distribuição dos pesos

A análise seguinte utiliza os pesos da convolução de expansão dos sete blocos. Ela permite verificar se a heterogeneidade das ativações acompanha diferenças nas magnitudes dos filtros ou se surge principalmente da combinação entre pesos, entradas, normalização e ativações.

````python
checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location="cpu",
    weights_only=True,
)

if "model_state_dict" in checkpoint:
    state_dict = checkpoint["model_state_dict"]
elif "state_dict" in checkpoint:
    state_dict = checkpoint["state_dict"]
else:
    state_dict = checkpoint

def find_tensor_by_suffix(suffix):
    matches = [
        (name, tensor)
        for name, tensor in state_dict.items()
        if name.endswith(suffix)
    ]
    if len(matches) != 1:
        raise KeyError(f"Expected one tensor ending with {suffix}, found {len(matches)}")
    return matches[0]

weight_records = []
weight_channel_records = []

for row in block_summary.itertuples(index=False):
    suffix = f"{row.block_name}.expand_1x1.convolution.weight"
    parameter_name, tensor = find_tensor_by_suffix(suffix)
    weights = tensor.detach().float().cpu().numpy()
    flattened_by_channel = weights.reshape(weights.shape[0], -1)

    for value in weights.reshape(-1):
        weight_records.append({
            "block": int(row.block),
            "weight": float(value),
        })

    activation_rows = (
        channel_statistics[channel_statistics["block"] == row.block]
        .sort_values("channel")
        .reset_index(drop=True)
    )
    if len(activation_rows) != weights.shape[0]:
        raise ValueError(
            f"Channel mismatch in block {row.block}: "
            f"{len(activation_rows)} activation channels and {weights.shape[0]} filters"
        )

    for channel, channel_weights in enumerate(flattened_by_channel):
        weight_channel_records.append({
            "block": int(row.block),
            "channel": channel,
            "weight_abs_max": float(np.max(np.abs(channel_weights))),
            "weight_mean_abs": float(np.mean(np.abs(channel_weights))),
            "weight_std": float(np.std(channel_weights)),
            "activation_scale": float(
                activation_rows.loc[channel, "channel_scale"]
            ),
        })

weight_distribution = pd.DataFrame(weight_records)
weight_channel_statistics = pd.DataFrame(weight_channel_records)
weight_channel_statistics.to_csv(
    TABLES_DIR / "expansion_weight_channel_statistics.csv",
    index=False,
)

display(
    weight_channel_statistics.groupby("block", as_index=False).agg(
        minimum_weight_abs_max=("weight_abs_max", "min"),
        maximum_weight_abs_max=("weight_abs_max", "max"),
        median_weight_abs_max=("weight_abs_max", "median"),
        median_weight_std=("weight_std", "median"),
    ).round(5)
)
````

````python
figure, axes = plt.subplots(1, 2, figsize=(11, 4.2))

selected_weight_distribution = weight_distribution[
    weight_distribution["block"].isin(selected_blocks)
]
sns.histplot(
    data=selected_weight_distribution,
    x="weight",
    hue="block",
    bins=70,
    stat="density",
    common_norm=False,
    element="step",
    fill=False,
    palette="tab10",
    ax=axes[0],
)
axes[0].set_xlabel("Expansion-convolution weight")
axes[0].set_ylabel("Density")
axes[0].set_title("(a) Weight distributions in grouped blocks")

for block, block_data in weight_channel_statistics.groupby("block"):
    selected = block in selected_blocks
    axes[1].scatter(
        block_data["weight_abs_max"],
        block_data["activation_scale"],
        s=28 if selected else 18,
        alpha=0.85 if selected else 0.3,
        label=f"Block {block}" if selected else None,
    )

valid_rows = weight_channel_statistics[
    (weight_channel_statistics["weight_abs_max"] > 0)
    & (weight_channel_statistics["activation_scale"] > 0)
]
log_weight = np.log10(valid_rows["weight_abs_max"].to_numpy())
log_activation = np.log10(valid_rows["activation_scale"].to_numpy())
log_correlation = float(np.corrcoef(log_weight, log_activation)[0, 1])

axes[1].set_xscale("log")
axes[1].set_yscale("log")
axes[1].set_xlabel("Maximum absolute weight per output channel")
axes[1].set_ylabel("Observed activation scale")
axes[1].set_title(f"(b) Weights versus activations, r = {log_correlation:.2f}")
axes[1].legend(frameon=False)

figure.tight_layout()
save_figure(figure, "figure_3_weight_distribution_and_activation_relation")
````

## 1.4 Relação entre diagnóstico e desempenho

A comparação final mostra se a intervenção direcionada aos canais heterogêneos recuperou o desempenho sem alterar o checkpoint treinado.

````python
test_results = results[results["split"] == "test"].copy()
model_order = [
    "original_fp32",
    "original_int8",
    "grouped_fp32",
    "grouped_int8",
]
model_labels = {
    "original_fp32": "Original\nFP32",
    "original_int8": "Original\nINT8",
    "grouped_fp32": "g8_4_4\nFP32",
    "grouped_int8": "g8_4_4\nINT8",
}
test_results["model"] = pd.Categorical(
    test_results["model"],
    categories=model_order,
    ordered=True,
)
test_results = test_results.sort_values("model")

figure, axis = plt.subplots(figsize=(7.2, 4.4))
bar_colors = ["#4C72B0", "#C44E52", "#55A868", "#8172B2"]
bars = axis.bar(
    [model_labels[str(model)] for model in test_results["model"]],
    test_results["accuracy"] * 100,
    color=bar_colors,
)
axis.axhline(5, color="#555555", linestyle="--", linewidth=1, label="Random chance (20 classes)")
axis.set_ylim(0, 105)
axis.set_ylabel("Test accuracy (%)")
axis.set_xlabel("Model")
axis.set_title("Effect of structural grouping on INT8 accuracy")
axis.legend(frameon=False, loc="lower right")

for bar, accuracy in zip(bars, test_results["accuracy"] * 100):
    axis.text(
        bar.get_x() + bar.get_width() / 2,
        accuracy + 1.5,
        f"{accuracy:.2f}%",
        ha="center",
        va="bottom",
        fontweight="bold",
    )

figure.tight_layout()
save_figure(figure, "figure_4_int8_accuracy_recovery")

accuracy_by_model = test_results.set_index("model")["accuracy"]
collapse_drop = float(
    accuracy_by_model["original_fp32"]
    - accuracy_by_model["original_int8"]
)
grouping_gain = float(
    accuracy_by_model["grouped_int8"]
    - accuracy_by_model["original_int8"]
)
recovered_fraction = grouping_gain / collapse_drop

diagnostic_summary = {
    "original_fp32_test_accuracy": float(accuracy_by_model["original_fp32"]),
    "original_int8_test_accuracy": float(accuracy_by_model["original_int8"]),
    "grouped_fp32_test_accuracy": float(accuracy_by_model["grouped_fp32"]),
    "grouped_int8_test_accuracy": float(accuracy_by_model["grouped_int8"]),
    "int8_collapse_percentage_points": collapse_drop * 100,
    "grouping_gain_percentage_points": grouping_gain * 100,
    "fraction_of_lost_accuracy_recovered": recovered_fraction,
    "weight_activation_log_correlation": log_correlation,
}

with (TABLES_DIR / "diagnostic_summary.json").open("w") as file:
    json.dump(diagnostic_summary, file, indent=2)

print(json.dumps(diagnostic_summary, indent=2))
````

## 1.5 Interpretação para o artigo

As evidências devem ser interpretadas em conjunto:

1. A razão entre escalas mostra a heterogeneidade interna de cada tensor.
2. O número aproximado de códigos mostra a resolução efetivamente disponível para cada canal sob uma escala compartilhada.
3. A partição g8_4_4 mostra como canais de magnitudes semelhantes foram isolados em tensores independentes.
4. A distribuição de pesos permite avaliar se a amplitude das ativações pode ser explicada somente pelos filtros.
5. A comparação de acurácia verifica se a intervenção estrutural é consistente com o diagnóstico.

A correlação entre pesos e ativações é descritiva e não estabelece causalidade. A evidência causal mais forte deste experimento é a comparação controlada entre o modelo original e o agrupado, ambos derivados do mesmo checkpoint e calibrados com o mesmo subconjunto estratificado.
